```python
import pandas as pd
import altair as alt
alt.data_transformers.disable_max_rows()
import numpy as np
```

```python
anime = pd.read_csv("../../data/preprocessed_anime.csv")
```

```python
anime['start_date'] = pd.to_datetime(anime['start_date'])
anime['end_date'] = pd.to_datetime(anime['end_date'])
```

```python
all_dates = []
for _, row in anime.iterrows():
    date_range = pd.date_range(row['start_date'], row['end_date'], freq='D')
    all_dates.extend(date_range)

# Convert to DataFrame and count occurrences
date_counts = pd.DataFrame(all_dates, columns=['date'])
date_counts = date_counts.groupby('date').size().reset_index(name='anime_count')

# Create the Altair chart
chart = alt.Chart(date_counts).mark_area().encode(
    x=alt.X('date:T', title='Date'),
    y=alt.Y('anime_count:Q', title='Active Anime Count'),
    tooltip=['date:T', 'anime_count:Q']
).properties(
    title="Number of Active Anime Over Time",
    width=1200,
    height=200
)
```

```python
anime['Score'] = pd.to_numeric(anime['Score'], errors='coerce')

# Generate a date range for each anime with scores
all_dates_scores = []
for _, row in anime.iterrows():
    date_range = pd.date_range(row['start_date'], row['end_date'], freq='D')
    all_dates_scores.extend([(date, row['Score']) for date in date_range])

# Convert to DataFrame and calculate average score per date
date_scores_df = pd.DataFrame(all_dates_scores, columns=['date', 'score'])
average_scores = date_scores_df.groupby('date').agg(
    avg_score=('score', 'mean'),  # Calculate average score
    anime_count=('score', 'count')  # Count active anime
).reset_index()

# Create Altair chart for average score over time
chart = alt.Chart(average_scores).mark_area().encode(
    x=alt.X('date:T', axis=alt.Axis(grid=False, title='Date')),  # X-axis: time
    y=alt.Y('avg_score:Q', 
            axis=alt.Axis(grid=False, title='Average Score'),
            scale=alt.Scale(domain=[4, 7.2], clamp=True)
            ),  # Y-axis: average score
    tooltip=['date:T', 'avg_score:Q', 'anime_count:Q']  # Tooltip: date, avg score, count
).properties(
    title="Average Anime Score Over Time with Count",
    width=1200,
    height=200
).configure_axis(
    grid=False  # Remove grid lines
)
```

```python
def generate_anime_count_by_date(df):
    all_dates = []
    
    df['start_date'] = pd.to_datetime(df['start_date'], errors='coerce')
    df['end_date'] = pd.to_datetime(df['end_date'], errors='coerce')
    
    df = df.dropna(subset=['start_date', 'end_date'])
    
    for _, row in df.iterrows():
        date_range = pd.date_range(row['start_date'], row['end_date'], freq='D')
        all_dates.extend(date_range)

    # Convert to DataFrame and count occurrences
    date_counts = pd.DataFrame(all_dates, columns=['date'])
    date_counts = date_counts.groupby('date').size().reset_index(name='anime_count')
    date_counts = date_counts.iloc[::30, :]

    # make date_counts to a array of x value and y value, convert y to year number
    date_counts = date_counts.to_numpy()
    date_counts = np.array(date_counts)
    date_counts = date_counts.tolist()
    
    # break to 2 arrays
    x = []
    y = []
    
    # convert x to string only keep date
    for i in date_counts:
        x.append(str(i[0])[:10])
        y.append(i[1])


    return x, y

x, y = generate_anime_count_by_date(anime)
```
